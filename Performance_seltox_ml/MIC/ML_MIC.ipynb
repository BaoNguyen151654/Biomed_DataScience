```python
import pandas as pd
import numpy as np

from sklearn.metrics import mean_squared_error, root_mean_squared_error, r2_score, mean_absolute_error

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import Lasso
from sklearn.svm import SVR
from xgboost import XGBRegressor

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
X_train = pd.read_csv('X_train_sel.csv')
X_test = pd.read_csv('X_test_sel.csv')
y_train = pd.read_csv('y_train_sel.csv')
y_test = pd.read_csv('y_test_sel.csv')

# Drop sn
X_train = X_train.drop('sn', axis=1)
X_test = X_test.drop('sn', axis=1)
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(y_train.squeeze(), bins=30, color='blue',
             label='Train', alpha=0.4, kde=True)

sns.histplot(y_test.squeeze(), bins=30, color='red',
             label='Test', alpha=0.4, kde=True)

plt.xlabel('MIC (µg/mL)')
plt.ylabel('Frequency')
plt.title('Distribution of y_train vs y_test')
plt.legend()
plt.grid(alpha=0.2)

plt.show()
```

```python
# Identify feature types
num_cols = X_train.select_dtypes(include=np.number).columns.tolist()
cat_cols = X_train.select_dtypes(exclude=np.number).columns.tolist()
```

```python
# Preprocessing
numeric_transformer = SimpleImputer(strategy='median')

categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])
preprocessor = ColumnTransformer([
    ('num', numeric_transformer, num_cols),
    ('cat', categorical_transformer, cat_cols)
])
```

```python
models = {
    'Random Forest': RandomForestRegressor(
        n_estimators=300,
        max_depth=None,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=-1
    ),

    'Lasso': Lasso(
        alpha=0.1,
        max_iter=10000
    ),

    'XGBoost': XGBRegressor(
        n_estimators=300,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        n_jobs=-1
    ),

    'SVR': SVR(
        kernel='rbf',
        C=10,
        epsilon=0.1,
        gamma='scale'
    )
}
```

```python
results = {}
predictions = {}

for name, model in models.items():

    pipeline = Pipeline([
        ('preprocessing', preprocessor),
        ('model', model)
    ])

    pipeline.fit(X_train, y_train)

    pred = pipeline.predict(X_test)

    # Ensure positive MIC predictions
    pred = np.maximum(pred, 0.000001)

    results[name] = pipeline
    predictions[name] = pred

    print(f"{name}: Training completed")
```

```python
evaluation = []

for name, pred in predictions.items():

    evaluation.append({
        'Model': name,
        'MAE': mean_absolute_error(y_test, pred),
        'MSE': mean_squared_error(y_test, pred),
        'RMSE': np.sqrt(mean_squared_error(y_test, pred)),
        'R2': r2_score(y_test, pred)
    })

evaluation_df = pd.DataFrame(evaluation)
evaluation_df = evaluation_df.sort_values('RMSE', ascending=True)
print(evaluation_df)
```

```python
y_actual = y_test.squeeze().values if hasattr(y_test, 'squeeze') else np.array(y_test)
sorted_indices = np.argsort(y_actual)
y_actual_sorted = y_actual[sorted_indices]

fig, axes = plt.subplots(2, 2, figsize=(16, 12))
axes = axes.ravel()

for idx, (name, pred) in enumerate(predictions.items()):
    ax = axes[idx]
    pred_sorted = np.array(pred)[sorted_indices]

    ax.plot(y_actual_sorted, label='Actual MIC', linestyle='-', linewidth=2, alpha=0.8)
    ax.plot(pred_sorted, label='Predicted MIC', linestyle='--', linewidth=1.5, alpha=0.8)

    ax.set_title(f'{name}: Actual vs Predicted MIC across Test Set')
    ax.set_xlabel('Test Samples (sorted by actual MIC)')
    ax.set_ylabel('MIC (µg/mL)')
    ax.legend()
    ax.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# Set up the subplot grid
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
axes = axes.ravel()

y_actual = y_test.squeeze()

for idx, (name, pred) in enumerate(predictions.items()):
    ax = axes[idx]
    # Plot actual vs predicted values
    ax.scatter(y_actual, pred, alpha=0.5, edgecolors='k', label='Predicted Data')
    # Plot the ideal line (y = x)
    lims = [
        np.min([ax.get_xlim(), ax.get_ylim()]),
        np.max([ax.get_xlim(), ax.get_ylim()])
    ]
    ax.plot(lims, lims, 'r--', alpha=0.75, zorder=3, label='Ideal Line (y = x)')
    # Plot the actual regression line
    m, b = np.polyfit(y_actual, pred, 1)
    ax.plot(y_actual, m * y_actual + b, linestyle='-', linewidth=2, label='Regression Line')
    ax.set_xlabel('Actual MIC (µg/mL)')
    ax.set_ylabel('Predicted MIC (µg/mL)')
    ax.set_title(f'{name}: Actual vs Predicted')
    ax.legend()
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
evaluation_df.to_csv('overallresult_ml_sel_MIC.csv', index=False)
```

```python
quantiles = np.percentile(y_actual, [25, 50, 75])

def get_quantile_group(val):
    if val <= quantiles[0]:
        return 'Q1 (0-25%)'
    elif val <= quantiles[1]:
        return 'Q2 (25-50%)'
    elif val <= quantiles[2]:
        return 'Q3 (50-75%)'
    else:
        return 'Q4 (75-100%)'

groups = np.array([get_quantile_group(v) for v in y_actual])

quantile_evaluation = []

for name, pred in predictions.items():
    for group in ['Q1 (0-25%)', 'Q2 (25-50%)', 'Q3 (50-75%)', 'Q4 (75-100%)']:
        mask = (groups == group)
        if np.sum(mask) > 0:
            y_true_g = y_actual[mask]
            y_pred_g = np.array(pred)[mask]

            mae = mean_absolute_error(y_true_g, y_pred_g)
            mse = mean_squared_error(y_true_g, y_pred_g)
            rmse = np.sqrt(mean_squared_error(y_true_g, y_pred_g))
            r2 = r2_score(y_true_g, y_pred_g) if len(y_true_g) > 1 else np.nan

            mic_min = np.min(y_true_g)
            mic_max = np.max(y_true_g)
            mic_range = f"{mic_min:.2f} - {mic_max:.2f}"

            quantile_evaluation.append({
                'Model': name,
                'Quantile': group,
                'N': np.sum(mask),
                'MIC range': mic_range,
                'MAE': mae,
                'MSE': mse,
                'RMSE': rmse,
                'R2': r2
            })

quantile_eval_df = pd.DataFrame(quantile_evaluation)
display(quantile_eval_df)
```

```python
quantile_eval_df.to_csv('quantileresult_ml_sel_MIC.csv', index=False)
```
